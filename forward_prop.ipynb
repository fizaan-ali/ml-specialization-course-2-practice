### ***Forward Prop in Numpy***

```python
def sigmoid(z):
    g = 1 / (1 + np.exp(z))
    return g
```

```python
def dense(a_in, W, b):
    units = W.shape[1]
    a_out = np.zeros(units)
    for j in range(units):
        w = W[:, j]
        z = np.dot(w, a_in) + b[j]
        a_out[j] = sigmoid(z)
    return a_out
```

```python
def sequential(x, W1, b1, W2, b2):
    a1 = dense(x, W1, b1)
    a2 = dense(a1, W2, b2)
    return a2
```

```python
def predict(X, W1, b1, W2, b2):
    m = X.shape[0]
    p = np.zeros((m,1))
    for i in range(m):
        p[i, 0] = sequential(X[i], W1, b1, W2, b2).item()
    return p
```

```python
W1_tmp = np.array( [[-8.93,  0.29, 12.9 ], [-0.1,  -7.32, 10.81]] )
b1_tmp = np.array( [-9.82, -9.28,  0.96] )
W2_tmp = np.array( [[-31.18], [-27.59], [-32.56]] )
b2_tmp = np.array( [15.41] )
```

```python
X_tst = np.array([
    [200,13.9],  # postive example
    [200,17]])   # negative example
X_tstn = norm_l(X_tst)  # remember to normalize
predictions = predict(X_tstn, W1_tmp, b1_tmp, W2_tmp, b2_tmp)
```

```python
yhat = (predictions >= 0.5).astype(int)
print(f"Decisions are: \n{yhat}")
```

```python
def dense_vec(X, W, B):
    '''
    These are dimesnions in this context
    X -> (1,2) 2D
    W -> (2,3) 2D
    B -> (1,3) 2D
    '''
    Z = np.matmul(X, W) + B # result will be (1,3)
    a_out = sigmoid(Z)
    return a_out
```
