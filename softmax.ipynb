```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.layers import Dense
from tensorflow.keras.models import Sequential
from sklearn.datasets import make_blobs
```

```python
def my_softmax(z):
    ez = np.exp(z)
    sm = ez / np.sum(ez)
    return sm
```

```python
# make  dataset for example
centers = [[-5, 2], [-2, -2], [1, 2], [5, -2]]
X_train, y_train = make_blobs(n_samples=2000, centers=centers, cluster_std=1.0,random_state=30)
```

```python
model = Sequential(
    [
        Dense(units=25, activation='relu'),
        Dense(units=15, activation='relu'),
        Dense(units=4, activation='softmax')
    ]
)
model.compile(
    loss = tf.keras.losses.SparseCategoricalCrossentropy(),
    optimizer = tf.keras.optimizers.Adam(0.001)
)
model.fit(X_train, y_train, epochs=10)
```

```python
p_nonpreferred = model.predict(X_train)
print(p_nonpreferred [:2]) # probabilites of first two examples 
print(f"largest value {np.max(p_nonpreferred):0.8f}, smallest value {np.min(p_nonpreferred):0.8f}")
```

```python
print(p_nonpreferred) # contains probabilites of 4 classes of all training examples
```

```python
preferred_model = Sequential(
    [
        Dense(units=25, activation='relu'),
        Dense(units=15, activation='relu'),
        Dense(units=4, activation='linear')
    ]
)
preferred_model.compile(
    loss = tf.keras.losses.SparseCategoricalCrossentropy( from_logits=True ), ####
    optimizer = tf.keras.optimizers.Adam(0.001)
)
preferred_model.fit(X_train, y_train, epochs=10)
```

```python
p_preferred = preferred_model.predict(X_train)
print(f"two example output vectors:\n{p_preferred[:2]}")
print(f"largest value: {np.max(p_preferred)}, smallest value: {np.min(p_preferred)}")
```

```python
sm_preferred = tf.nn.softmax(p_preferred).numpy()
print(f"two example output vectors:\n{sm_preferred[:2]}")
print(f"largest value: {np.max(sm_preferred)}, smallest value: {np.min(sm_preferred)}")
```

```python
# can also find index of largest value using argmax without having to use softmax func
for i in range(5):
    print(f"{p_preferred[i]}, category: {np.argmax(p_preferred[i])}")
```
