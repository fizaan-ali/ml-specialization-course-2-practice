```python
import numpy as np
```

```python
AT = np.array([[200, 17]])
W = np.array([[1, -3, 5],
             [-2, 4, -6]])
B = np.array([[-1, 1, 2]])
```

```python
def sigmoid(z):
    g = 1 / (1 + np.exp(-z))
    return g
```

```python
def dense_vec(AT, W, B):
    Z = np.matmul(AT, W) + B # or Z = AT @ W
    A_out = sigmoid(Z)
    return A_out
```
