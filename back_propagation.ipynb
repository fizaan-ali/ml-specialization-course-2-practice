```python
from sympy import *
import numpy as np
import matplotlib.pyplot as plt
import re
```

```python
w = 3
a = 2+3*w
J = a**2
print(f"w = {w}, a = {a}, J = {J}")
```

```python
a_epsilon = a + 0.001
J_epsilon = a_epsilon**2
k = (J_epsilon - J) / 0.001
print(f"J = {J}, J_epsilon = {J_epsilon}, dJ_da ~= k = {k}")
```

```python
sw, sJ, sa = symbols('w, J, a')
sJ = sa**2
sJ
```

```python
sJ.subs([(sa, a)])
```

```python
dJ_da = diff(sJ, sa)
dJ_da
```

```python
dJ_da.subs([(sa, a)])
```

```python
w_epsilon = w + 0.001
a_epsilon = 2 + 3 * w_epsilon
k = (a_epsilon - a) / 0.001
print(f"a = {a}, a_epsilon = {a_epsilon}, da_dw ~= k = {k}")
```
